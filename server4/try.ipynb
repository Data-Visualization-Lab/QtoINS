```python
import contextlib
import io
from function.SQL2NL_clean import sql2nl

SQL = """SELECT Title,'IMDB Rating','Rotten Tomatoes Rating'
FROM movies
WHERE Genre In ('Comedy','Black Comedy',
                'Romantic Comedy')
      AND 'Release Year' between 2002 and 2010"""  # input your SQL here


raw_explanation_data = sql2nl(SQL)

def to_full_sentence(data):
    if not data:
        return ""

    query_sentences = []
    for query in data:
        steps = []
        for item in query.get("explanation", []):
            text = item.get("explanation", "").strip()
            if not text:
                continue
            steps.append(text.rstrip("."))

        if steps:
            sentence = ", then ".join(steps)
            sentence = sentence[0].upper() + sentence[1:]
            if not sentence.endswith("."):
                sentence += "."
            query_sentences.append(sentence)

    return " ".join(query_sentences)

explanation_data = to_full_sentence(raw_explanation_data)
explanation_data
```
